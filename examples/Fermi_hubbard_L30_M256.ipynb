# MTBL: Fermi–Hubbard example (30 sites)

**MTBL means Modified Thick-Block Lanczos.** The historical callable `trl_variance_solver` is the residual-enriched restart routine used in this example.

Reference: [An accurate Lanczos method for the matrix product state representation](https://doi.org/10.1038/s42005-026-02896-x), *Communications Physics* **9**, 326 (2026).

This preserves the parameters of the supplied research experiment. It can be computationally expensive; run `quickstart.ipynb` first. Full-scale convergence is not validated by the smoke tests.

Install from the repository root with `python -m pip install -e ".[examples]"`, then select that environment's Jupyter kernel. All functions are defined in the Python package; this notebook contains only example calls. See [algorithm notes](../docs/algorithms.md) and the [API reference](../docs/api.md).

```python
import matplotlib.pyplot as plt
from restart_lanczos_mps import trl_variance_solver
from restart_lanczos_mps.examples import make_hubbard_problem, plot_variance_history
```

## Configure the example

The seed is fixed for reproducibility. As in the original notebooks, quantum numbers are disabled; this example does **not** enforce a particle-number or half-filling sector. `tol` is a threshold on the **squared compressed residual norm** in this implementation.

```python
nsites = 30
max_bond = 256
k = 4
seed = 42
initial_lanczos_steps = 30
num_restarts = 1000
lanczos_steps = 4
tol = 1e-06
trunc_tol = 1e-10
```

## Construct the Hamiltonian and initial state

```python
hamiltonian, initial_state = make_hubbard_problem(
    nsites=nsites, interaction=8.0, hopping=1.0, chemical_potential=0.0, seed=seed
)
```

## Run the MTBL example

```python
eigenvalues, eigenvectors, variance_history = trl_variance_solver(
    hamiltonian, k, max_bond, initial_state,
    initial_lanczos_steps=initial_lanczos_steps,
    num_restarts=num_restarts,
    lanczos_steps=lanczos_steps,
    tol=tol,
    trunc_tol=trunc_tol,
)
print("Returned Ritz values:", eigenvalues)
```

## Inspect convergence

The horizontal axis counts recorded restart cycles, not MPO applications. Values are plotted without energy rescaling. Locked-slot zero placeholders are masked on the logarithmic plot. The compressed residual diagnostic need not equal the exact energy variance.

```python
ax = plot_variance_history(variance_history)
plt.show()
```
